```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
pass_path = "../data/labeled_data/0_PASS/0_PASS.xlsx"
fail_path = "../data/labeled_data/1_FAIL/1_FAIL.xlsx"
# un_path = "../data/labeled_data/1_FAIL/1_FAIL.xlsx"

df_pass = pd.read_excel(pass_path)
df_fail = pd.read_excel(fail_path)
# df_un = pd.read_excel(un_path)
```

```python
random_state = 42
pass_train, pass_test = train_test_split(
    df_pass, 
    test_size = 0.2,
    random_state = random_state
)
```

```python
fail_train, fail_test = train_test_split(
    df_fail, 
    test_size = 0.2,
    random_state = random_state
)
```

```python
un_train, un_test = train_test_split(
    df_un, 
    test_size = 0.2,
    random_state = random_state
)
```

```python
df_train = pd.concat([pass_train, fail_train], ignore_index=True)
df_test = pd.concat([pass_test, fail_test], ignore_index=True)
```

```python
df_train = df_train.sample(frac=1, random_state=random_state).reset_index(drop=True)
df_test = df_test.sample(frac=1, random_state=random_state).reset_index(drop=True)
```

```python
import os
def save_or_update(df, file_path):
    folder = os.path.dirname(file_path)
    if folder:
        os.makedirs(folder, exist_ok=True)
    df.to_excel(file_path, index=False)
```

```python
train_path = "../data/labeled_data/train/train.xlsx"
test_path = "../data/labeled_data/test/test.xlsx"
```

```python
save_or_update(df_train, train_path)
```

```python
save_or_update(df_test, test_path)
```
